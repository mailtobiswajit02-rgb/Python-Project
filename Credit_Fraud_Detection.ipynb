```python
# SMOTE Technique

# Synthetic Minority Over Sampling Technique
```

```python
# Why do we need SMOTE ?

# Our dataset is highly imbalanced , 99.83 / 0.16

# Problem:

# Model learns mostly the genuine transactions
# Ignore the fraud cases
# Gives very high accuracy but poor fraud detection

# How SMOTE works:

# Picks a minority class point(Fraud Class)
# Finds its nearest neighbours
# Creates a new point between them

# 120 , 130 , 125

# *** It balances an imbalaced dataset
# *** By generting synethetic samples
```

```python
!pip install imbalanced-learn
```

```python
from imblearn.over_sampling import SMOTE
```

```python
smote = SMOTE(random_state = 42)
```

```python
print(y_train.value_counts())
```

```python
x_train_smote,y_train_smote= smote.fit_resample(x_train,y_train)
```

```python
print("Before SMOTE:" ,y_train.value_counts())
print("After SMOTE: ", y_train_smote.value_counts())
```

```python
# We will use a Decision Tree
# This prob involves making a decision based on multiple features

# if age > 30 -> check the income -> check credit score -> approve or reject
```

```python
# Model Building

# Decision Tree
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
dt_model = DecisionTreeClassifier()
```

```python
# Train the model
dt_model.fit(x_train_smote,y_train_smote)
```

```python
# Test the model
dt_pred = dt_model.predict(x_test)
```

```python
dt_pred
```

```python
from sklearn.metrics import *
```

```python
accuracy_score(y_test,dt_pred)
```

```python
precision_score(y_test,dt_pred)
```

```python
recall_score(y_test,dt_pred)
```

```python
print(classification_report(y_test,dt_pred))
```

```python
print("Without SMOTE")
dt_model.fit(x_train,y_train)
print(classification_report(y_test,dt_model.predict(x_test)))


print("With SMOTE")
dt_model.fit(x_train_smote,y_train_smote)
print(classification_report(y_test,dt_model.predict(x_test)))
```

```python
# Precision - When model predicted fraud, how often is it correct
# Recall - Out of all the fraud cases, how many did the model catch?

# Precision - reference is prediction...
# Recall - reference is Actual Fraud cases
```

```python
# To detect as many fraud transactions as possible
# While keeping false alarms at an accepetance level
```
