Separate movie if from customer id

```python
movie_id = None # movie id is a variable which will iterate through all the movie ids.
movie_id_per_row = []

for i in netflix_dataset['CustId']:
  if ":" in i:
    movie_id = int(i.replace(":","")) # store the value of movie id without colon
  movie_id_per_row.append(movie_id)
```

## Create a list of movies with very few ratings

```python
summary = netflix_dataset.groupby("Movie_ID")['Rating'].agg(['count'])
summary
```

```python
# now we will keep a benchmark at 60th percentile

movie_benchmark = round(np.percentile(summary['count'], 60, method ='linear'))
```

```python
movie_benchmark
```

```python
drop_movie_list = summary[summary['count']<movie_benchmark].index
```

```python
drop_movie_list
```

```python
summary_cust =  netflix_dataset.groupby("CustId")['Rating'].agg(['count'])
summary_cust
```

```python
cust_benchmark = round(np.percentile(summary_cust['count'], 60, method='linear'))
```

```python
cust_benchmark
```

```python
drop_cust_list = summary_cust[summary_cust['count']<cust_benchmark].index
drop_cust_list
```

```python
netflix_dataset = netflix_dataset[~netflix_dataset['Movie_ID'].isin(drop_movie_list)]

# so it checks all the movies that are present in the drop movie list using membership operator
# then using negation operator ~ we filter out all those rows
# finally your data has all the rows with the movie ids not present in the drop movie list.
```

```python
netflix_dataset = netflix_dataset[~netflix_dataset['CustId'].isin(drop_cust_list)]
```
