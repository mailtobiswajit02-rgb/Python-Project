# **Prophet Library:**


Prophet is an additive time series Forecasting models.


It breaks the time series into components

Main Equation:

y(t)=g(t)+s(t)+h(t)+et

g(t): Trend Component

s(t): Seosnal Component

h(t): Holiday/Event Component

et= Error component

# **How is Prophet different from Arima/Sarima:**

Feature                        Arima         Sarima                 Prophet

Needs Stationary Data          Yes           Yes                      No

Differencing                  Yes            Yes                      No

Parameters                    p,d,q          p,d,q,P,D,Q,m           Automatic

Missing Values               Poor            Poor                   Good

Holiday Effect                No              NO                     Yes




# **Disadvantages of Prophet:**

1. Limited performance on highly complex time dependcies:

2. Assumes additive struture.

3. Less control over statistical modelling.

# Forecasting for Confirmed Cases

```python
#Rename the column in confirmed data to ds and y
confirmed.columns=['ds','y']
```

```python
confirmed
```

```python
#Import the Library
from prophet import Prophet
```

```python
model=Prophet()
```

```python
#Train the Model
model.fit(confirmed)
```

```python
#Make prediction for next 7 days
future=model.make_future_dataframe(periods=7)
future
```

```python
forecast=model.predict(future)
```

```python
forecast
```

Columns in the output:


ds: The timestamp for which preditions are made

yhat: The predicted value(forecast output) for the corresponding date.

yhat lower and yhat upper:

The lower and upper bounds of the uncertainity intervals for the forecast

trend: The estimated trend component oof the forecast.

```python
#Visulaise the forecst
model.plot(forecast)
plt.show()
```
